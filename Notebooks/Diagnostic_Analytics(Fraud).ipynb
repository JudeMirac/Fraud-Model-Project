```python
# Import dataset

import pandas as pd 

df = pd.read_csv('fraud_risk_dataset.csv')

#view data 
df.head()
```

```python
# In this notebook we will begin look at the why fraud occurs 
```

```python
# Test 1 

df.groupby('international')['fraud'].mean()
```

```python
# Foreign transactions see an increase in the amount of fraud 
```

```python
# Test 2
df.groupby('num_transactions_24h')['fraud'].mean()
```

```python
# There is a increase of recorded fraud on transactions that are gretater than 5
```

```python
#Test 3

df.groupby('previous_chargebacks')['fraud'].mean() 
```

```python
# Repeated offenders of charge backs = reapted fraud
```

```python
# test 4 
df.groupby('device_type')['fraud'].mean()
```

```python
# Tablets and mobile phone pose a huge risk for fraud 
```

```python
#Test 5 
df['age_group'] = pd.cut(
    df['customer_age'],
    bins = [18, 25, 35, 45, 55, 65, 100])

df.groupby('age_group')['fraud'].mean()
```

```python
# Age between 18 - 25 pose the highest risk for fraudulent transactions, second would be 25 - 35. 
# Young adults are likely to commit fraudulent transactions 
```

```python
# Test 6 
pd.crosstab(df['payment_method'], df['international'], values = df['fraud'], aggfunc='mean')
```

```python
# The highest ammount of fraud on average is being done by foreigners using credit transactions 
```

```python
df['velocity'] = pd.cut(
    df['num_transactions_24h'], bins = [2,5,8,10]
)

df.groupby('velocity')['fraud'].mean()
```

```python
# Test 7 

df.corr(numeric_only=True)   
```

```python
# Fraud likelihood increases sharply for international transactions and customers with chargeback history.
# Mobile devices show higher fraud rates compared to desktops or tablets.
# Transaction velocity strongly elevates risk up to medium-high levels.
# High-value categories such as luxury show persistent fraud exposure.
# Geography and payment behavior are superior predictors over demographics.
```
